```python
# Group anagrams from list: ["eat", "tea", "tan", "ate", "nat", "bat"]

words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for char in words:

    key = "".join(sorted(char))

    if key not in groups:
        groups[key] = []

    groups[key] = char


print(list(groups.values()))
```

```python
# Find all starting indices where an anagram of "abc" appears in "cbaebabacd"


pattern = sorted("abc")

word = "cbaebabacd"

result = []

for i in range(len(word) - len(pattern) +1):

    window = word[i : i + len(pattern)]

    if sorted(window) == pattern:
        result.append(i)

print(result)
```

```python
# Find the FIRST non-repeating character in "aabbccdeef"

word = "aabbccdeef"

freq = {}



for char in word:    
    freq[char] = freq.get(char,0) + 1

for char in word:
    if freq[char] == 1:
        print(char)
        break
    
```

```python
# Find the MOST COMMON character in "aabbbbcccc"

word = "aabbbbcccc"

freq = {}

for char in word:
    freq[char] = freq.get(char,0) + 1

max_value = max(freq.values())

result = []

for key in freq:
    if freq[key] == max_value:
        result.append(key)

print(result)
        
```

```python
# Check if all characters in "aabbcc" appear EVEN number of times.

word = "aabbcc"

freq = {}

for char in word:
    freq[char] = freq.get(char,0) + 1

result = True

for key in freq:
    if freq[key] % 2 != 0:
        result = False
        break

print(result)
```

```python
# Count how many characters appear EXACTLY ONCE in "programming"

word = "programming"

freq = {}

for char in word:
    freq[char] = freq.get(char,0) + 1

count = 0

for key in freq:
    
    if freq[key] == 1:
        count += 1

print(count)
```

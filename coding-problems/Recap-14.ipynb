```python
# Find all starting positions where "ab" appears in "ababcab"

word = "ababcab"

pattern = "ab"

result = []

for i in range(len(word) - len(pattern) + 1):

    window = word[i : i + len(pattern)]

    if window == pattern:
        result.append(i)

print(result)
```

```python
# Check if "hello world" contains the substring "wor"

word = "hello world"

pattern = "wor"

result = False

for i in range(len(word) - len(pattern) + 1):

    window = word[i : i + len(pattern)]

    if window == pattern:
        result = True
        break

print(result)
```

```python
# Count how many times "is" appears in "this is a test, is it?"

word = "this is a test, is it?"

pattern = "is"

count = 0

for i in range(len(word) - len(pattern) + 1):

    window = word[i : i + len(pattern)]

    if window == pattern:
        count += 1

print(count)
```

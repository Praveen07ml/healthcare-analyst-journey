# Two Pointers - Mix Patterns 

```python
# Check if [1, 2, 3, 4, 5] is sorted in ascending order

numbers = [1, 2, 3, 4, 5]

result = True


for i in range(1,len(numbers)):
    if numbers[i] < numbers[i-1]:
        result = False
        break

if result:
    print("Sorted Array")
else:
    print("Not Sorted Array")
        
        
```

```python
# In SORTED array [1, 5, 3, 7, 9], find pair with DIFFERENCE = 4.

numbers = [1, 5, 3, 7, 9]

target = 4

result = []

left = 0
right = len(numbers) -1

while left < right:
    difference = numbers[right] - numbers[left]

    if difference == target:
        result.append([numbers[left],numbers[right]])
        
        left += 1
        right -= 1
    elif difference < target:
        right -= 1
    else:
        left += 1

print(result)
```

```python
# Merge two SORTED lists [1, 3, 5] and [2, 4, 6] into [1, 2, 3, 4, 5, 6]

list1 = [1, 3, 5]
list2 = [2, 4, 6]

result = []

i = 0
j = 0

while i < len(list1) and j < len(list2):

    if list1[i] < list2[j]:
        result.append(list1[i])
        i += 1
    else:
        result.append(list2[j])
        j += 1

while i < len(list1):
    result.append(list1[i])
    i += 1

while j < len(list2):
    result.append(list2[j])
    j += 1

print(result)
    
```

```python
# Find intersection (common elements) of [1, 2, 3, 4, 5] and [3, 4, 5, 6, 7]

list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

result = []

for i in range(len(list1)):
    if list1[i] in list2:
        result.append(list1[i])

print(result)
```

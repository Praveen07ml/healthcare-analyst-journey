```python
# Move all NEGATIVE numbers to left, POSITIVE to right: [-1, 2, -3, 4, -5, 6]

numbers = [-1, 2, -3, 4, -5, 6]

slow = 0

for i in range(len(numbers)):

    if numbers[i] < 0:
        numbers[slow],numbers[i] = numbers[i],numbers[slow]
        slow += 1

print(numbers)
```

```python
# Partition array [7, 2, 5, 3, 8, 1, 4] around pivot value 4.

numbers = [7, 2, 5, 3, 8, 1, 4]

slow = 0
for i in range(len(numbers)):

    if numbers[i] <= 4:

        numbers[i],numbers[slow] = numbers[slow],numbers[i]
        slow += 1

print(numbers)
```

```python
# Sort array containing only 0s and 1s: [1, 0, 1, 1, 0, 0, 1]

numbers = [1, 0, 1, 1, 0, 0, 1]

left = 0
for i in range(len(numbers)):

    if numbers[i] <= 0:
        numbers[i],numbers[left] = numbers[left],numbers[i]
        left += 1

print(numbers)
```

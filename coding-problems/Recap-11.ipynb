```python
# Reverse the string "hello world" to get "dlrow olleh"

word = "hello world"

result = ""

for i in range(len(word)-1,-1,-1):
    result += word[i]

print(result)
    
```

```python
# Reverse only the WORDS in "hello world python" to get "python world hello"

words = "hello world python"

words = words.split()

result = []

for i in range(len(words)-1,-1,-1):

    result.append(words[i])

print(" ".join(result))

    
```

```python
# Rotate string "abcdef" LEFT by 2 positions to get "cdefab"

word = "abcdef"

target = 2

result = ""

for i in range(len(word)):
    
    if target == i:
        result += word[target:]

result += word[:target]

print(result)
        
```

```python
# Reverse only VOWELS in "hello" to get "holle"

word = "hello"

word = list(word)
vowels = "aeiouAEIOU"

left = 0
right = len(word) -1

while left < right:

    if word[left] not in vowels:
        left += 1

    elif word[right] not in vowels:
        right -= 1

    else:
        word[left],word[right] = word[right],word[left]
        left += 1
        right -= 1

print("".join(word))
```

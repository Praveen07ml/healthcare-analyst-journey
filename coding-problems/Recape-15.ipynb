```python
# Find the largest number in [34, 12, 78, 23, 90, 45, 67]

numbers = [34, 12, 78, 23, 90, 45, 67]

maximum = numbers[0]


for number in numbers:
    if number > maximum:
        maximum = number

print(maximum)
```

```python
# Find the SMALLEST number in [45, 12, 8, 67, 23, 5, 34]

numbers = [45, 12, 8, 67, 23, 5, 34]

minimum = numbers[0]

for number in numbers:
    if number < minimum:
        minimum = number

print(minimum)
```

```python
# Find the largest EVEN number in [15, 22, 8, 45, 18, 33, 7]

numbers = [15, 22, 8, 45, 18, 33, 7]

even_large = numbers[0]

for number in numbers:
    if number % 2 == 0 and number > even_large:
        even_large = number

print(even_large)
```

```python
# Reverse [1, 2, 3, 4, 5] to get [5, 4, 3, 2, 1]

numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) -1

while left < right:

    numbers[left],numbers[right] = numbers[right],numbers[left]

    left += 1
    right -= 1

print(numbers) 
```

```python
# Reverse the string "hello" to get "olleh"

word = "hello"

result = ""

for i in range(len(word)-1,-1,-1):
    result += word[i]

print(result)
```

```python
# Find the second largest number in [34, 12, 78, 23, 90, 45, 67]


numbers = [34, 12, 78, 23, 90, 45, 67]

if len(numbers) < 2:
    print(None)

else:

    largest = numbers[0]
    second_largest = numbers[1]

    for i in range(len(numbers)):

        element = numbers[i]

        if element > largest:
            second_largest = largest
            largest = element

        elif element > second_largest and element != largest:
            second_largest = element

    print(second_largest)
    
```

```python
# Find the THIRD largest in [34, 12, 78, 23, 90, 45, 67]

numbers  =  [34, 12, 78, 23, 90, 45, 67]

if len(numbers) < 3:
    print(None)

else:

    largest = numbers[0]
    s_largest = numbers[1]
    t_largest = numbers[2]

    for i in range(len(numbers)):

        element = numbers[i]

        if element > largest:
            t_largest = s_largest
            s_largest = largest
            largest = element

        elif element > s_largest and element != largest:
            t_largest = s_largest
            s_largest = element

        elif element > t_largest and element != s_largest and element != largest:
            t_largest = element

    print(t_largest)
    print(s_largest)
    print(largest)
            
```

```python
# Find the MEDIAN (middle value) of [34, 12, 78, 23, 90]

numbers = [34, 12, 78, 23, 90]

numbers.sort()

if len(numbers) % 2 == 0:
    print(numbers[len(numbers)//2])

else:
    print(numbers[len(numbers)//2-1] + numbers[len(numbers)//2] /2)
```

```python
# Remove all zeros from [0, 1, 0, 3, 0, 12] to get [1, 3, 12]


numbers = [0, 1, 0, 3, 0, 12]

result = []

for number in numbers:

    if number != 0:
        result.append(number)

print(result)
```

```python
# Remove duplicates from [1, 2, 2, 3, 4, 4, 5] to get [1, 2, 3, 4, 5]

numbers = [1, 2, 2, 3, 4, 4, 5] 

seen = set()
result = []

for number in numbers:
    if number not in seen:
        seen.add(number)
        result.append(number)

print(result)
```

```python
# Merge [1, 4, 7] and [2, 3, 5, 8, 9] into one sorted list


num1 = [1, 4, 7] 
num2 = [2, 3, 5, 8, 9]

i = 0
j = 0

result = []

while i < len(num1) and j < len(num2):

    if num1[i] < num2[j]:
        result.append(num1[i])
        i += 1

    elif num1[i] > num2[j]:
        result.append(num2[j])
        j += 1

while i < len(num1):
    result.append(num1[i])
    i += 1

while j < len(num2):
    result.append(num2[j])
    j += 1

print(result)
        
```

```python
# Combine [1, 2, 3] and [4, 5, 6] by alternating elements: [1, 4, 2, 5, 3, 6]

num1 = [1, 2, 3]
num2 = [4, 5, 6]

result = []

i = 0
j = 0

while i< len(num1) and j < len(num2):

    result.append(num1[i])
    result.append(num2[j])
    i +=1
    j += 1

print(result)
```

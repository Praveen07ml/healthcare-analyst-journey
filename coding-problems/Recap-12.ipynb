```python
# Check if "listen" and "silent" are anagrams

a = "listen" 
b = "silent"

if len(a) != len(b):
    print(False)

else:

    a = sorted(a)
    b = sorted(b)

    if a == b:
        print(True)
    else:
        print(False)
```

```python
# Check if "Hello" and "Olleh" are anagrams (ignore case)

a = "Hello"
b = "Olleh"

x = ""
y = ""

for char in a:
    x += char.lower()

for char in b:
    y += char.lower()

if len(x) != len(y):
    print(False)

else:

    freq = {}

    for char in x:
        freq[char] = freq.get(char,0) + 1

   
    
    for char in y:
        freq[char] = freq.get(char,0) - 1

    print(all( value == 0 for value in freq.values()))
    
```

```python
# Group anagrams from list: ["eat", "tea", "tan", "ate", "nat", "bat"]


words = ["eat", "tea", "tan", "ate", "nat", "bat"]


groups = {}

for word in words:

    key = "".join(sorted(word))

    if key not in groups:

        groups[key] = []

    groups[key].append(word)


print(list(groups.values()))
    




```

```python
# Find all starting indices where an anagram of "abc" appears in "cbaebabacd"

pattern = sorted("abc")

word = "cbaebabacd"

result = []

for i in range(len(word) - len(pattern) + 1):

    window = word[i : i + len(pattern)]

    if sorted(window) == pattern:
        result.append(i)

print(result)
```

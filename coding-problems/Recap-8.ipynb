```python
# Sort array containing only 0s and 1s: [1, 0, 1, 1, 0, 0, 1]


numbers = [1, 0, 1, 1, 0, 0, 1]

slow = 0 # for non zeros

for fast in range(len(numbers)):

    if numbers[fast] != 0:
        numbers[slow],numbers[fast] = numbers[fast],numbers[slow]
        slow += 1

print(numbers)
```

```python
# Move all NEGATIVE numbers to left, POSITIVE to right: [-1, 2, -3, 4, -5, 6]

numbers = [-1, 2, -3, 4, -5, 6]

slow = 0 # positives

for fast in range(len(numbers)):

    if numbers[fast] > 0 :
        numbers[slow],numbers[fast] = numbers[fast],numbers[slow]
        slow += 1

print(numbers)
```

```python
# Partition array [7, 2, 5, 3, 8, 1, 4] around pivot value 4.

numbers = [7, 2, 5, 3, 8, 1, 4]

slow = 0

for fast in range(len(numbers)):

    if numbers[fast] >= 4:

        numbers[slow],numbers[fast] = numbers[fast],numbers[slow]
        slow += 1

print(numbers)
```

```python
# Sort array with 0s, 1s, and 2s: [2, 0, 2, 1, 1, 0]


numbers = [2, 0, 2, 1, 1, 0]

low = 0
mid = 0
high = len(numbers) - 1

while mid < high:

    if numbers[low] == 0:
        numbers[low],numbers[mid] = numbers[mid],numbers[low]
        low += 1
        mid += 1

    elif numbers[mid] == 1
        numbers[mid],numbers[high] = numbers[high],numbers[mid]
        mid += 1

    else:
        numbers[high],numbers[mid] = numbers[mid],numbers[high]
        high -= 1

print(numbers)
```

```python
# Check if "A man, a plan, a canal: Panama" is a palindrome

words = "A man, a plan, a canal: Panama"

word = "" 

for char in words:
    if char.isalpha():
        word += char.lower()

left = 0
right = len(word) -1

result = True

while left < right:

    if word[left] != word[right]:
        result = False
        break

    left += 1
    right -= 1

if result:
    print("Palindrome")

else:
    print("Not palindrome")
        
```

```python
# Reverse only the VOWELS in "hello" to get "holle".

word = "hello"

word = list(word)

vowels = list("AEIOUaeiou")

left = 0
right = len(word) -1

while left < right:

    if word[left] not in vowels:
        left += 1

    elif word[right] not in vowels:
        right -= 1

    else:
        
        word[left],word[right] = word[right],word[left]
        
        left += 1
        right -= 1

print("".join(word))
```

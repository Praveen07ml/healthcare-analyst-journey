# String patterns mix

```python
# Compress "aaabbbccc" to "a3b3c3" (character + count format)

word = "aaabbbccc"

word = sorted(word)

result = ""

count = 1

for i in range(1,len(word)):

    if word[i] == word[i -1]:
        count += 1

    else:
        
        result += word[i-1] + str(count)
        count = 1

result += word[-1] + str(count)

print(result)
```

```python
# Check if "madam" is a palindrome.

word = "madam"

result = True

left = 0
right = len(word) - 1

while left < right:
    
    if word[left] != word[right]:
        result = False
        break

    left += 1
    right -= 1

print(result)
        
```

```python
# Find longest common prefix of ["flower", "flow", "flight"]

words = ["flower", "flow", "flight"]

prefix = ""

found = False

for i in range(len(words[0])):

    char = words[0][i]

    for word in words:

        if i >= len(word) or word[i] != char:
            found = True
            break

    if found:
        break

        
    prefix += char

print(prefix)
```

```python
# Remove all adjacent duplicate characters from "abbaca"


word = "abbaca"

word = list(word)

result = []

for i in range(1,len(word)):
    
    if word[i] != word[i - 1]:
        result.append(word[i])

print("".join(result))
```

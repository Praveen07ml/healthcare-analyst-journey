```python
# Check if [1, 2, 3, 4, 5] is sorted in ascending order.

numbers = [1, 2, 3, 4, 5]

result = True

for i in range(1,len(numbers)):
    if numbers[i] < numbers[i-1]:
        result = False
        break

print(result)
```

```python
# In SORTED array [1, 5, 3, 7, 9], find pair with DIFFERENCE = 4.
# (Difference means |a - b| = 4)


numbers = [1, 5, 3, 7, 9]

numbers.sort()

target = 4

result = []

left = 0
right = 1 

while right < len(numbers):
    
    diff = numbers[right] - numbers[left]

    if diff > target:
        left += 1
        
    elif diff < target:
        right += 1
        
    else:
        result.append([numbers[right],numbers[left]])
        left +=1
        right += 1

    if left == right:
        right += 1


print(result)    
```

```python
# Merge two SORTED lists [1, 3, 5] and [2, 4, 6] into [1, 2, 3, 4, 5, 6]

list1 = [1, 3, 5]
list2 = [2, 4, 6]


result = []

for i in range(len(list1)):
    if list1[i] < list2[i]:
        result.append(list1[i])
        result.append(list2[i])

print(result)
```

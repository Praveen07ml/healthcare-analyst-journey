# String Reversal and Rotation pattern

```python
# Reverse the string "hello world" to get "dlrow olleh"

word = "hello world"

# first Approach

result = ""

for i in range(len(word) -1,-1,-1):
    result = result + word[i]

print(result)


# Second Approach


word = list(word)

left = 0
right = len(word) -1

while left < right:
    word[left],word[right] = word[right],word[left]
    left += 1
    right -= 1

print("".join(word))
```

```python
# Reverse only the WORDS in "hello world python" to get "python world hello"

words = "hello world python"
words = words.split()

result = []

for i in range(len(words) - 1,-1,-1):
    result.append(words[i])

print(" ".join(result))
    
```

```python
# Reverse only the WORDS in "hello world python" to get "python world hello" using two pointer swap method

words = "hello world python"
words = words.split()

left = 0
right = len(words) -1

while left < right:
    words[left],words[right] = words[right],words[left]
    left += 1
    right -= 1

print(" ".join(words))
```

```python
# Reverse only VOWELS in "hello" to get "holle"

word = list("hello")
vowels = "aeiou"

left = 0
right = len(word) - 1

while left < right:
    if word[left] not in vowels:
        left += 1
    elif word[right] not in vowels:
        right -= 1
    else:
        word[left],word[right] = word[right],word[left]
        left += 1
        right -= 1

print("".join(word))
```

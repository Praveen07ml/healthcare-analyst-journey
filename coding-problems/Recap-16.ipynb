```python
# Check if list [1, 2, 3, 2, 1] is a palindrome (reads same forwards and backwards)

numbers = [1, 2, 3, 2, 1]

result = "Palindrome"

left = 0
right = len(numbers) - 1

while left < right:

    if numbers[left] != numbers[right]:
        result = "Not Palindrome"
        break

    left += 1
    right -= 1

print(result)
        
```

```python
# In SORTED array [1, 2, 3, 4, 6], find TWO numbers that sum to 6.

numbers = [1, 2, 3, 4, 6]

target = 6

result = []

left = 0
right = len(numbers) - 1

while left < right:

    total = numbers[left] + numbers[right]

    if total > target:
        right -= 1

    elif total < target:
        left += 1

    else:
        result.append([numbers[left],numbers[right]])
        left += 1
        right -= 1

print(result)
    
```

```python
# Reverse list [1, 2, 3, 4, 5] IN-PLACE using two-pointer swap.


numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) - 1

while left < right:

    numbers[left],numbers[right]=numbers[right],numbers[left]
    
    left += 1
    right -= 1
print(numbers)
```

```python
# Move all zeros to the end: [0, 1, 0, 3, 12] → [1, 3, 12, 0, 0]

numbers = [0, 1, 0, 3, 12]

slow = 0

for i in range(len(numbers)):

    if numbers[i] != 0:
        numbers[slow] = numbers[i]
        slow += 1

for i in range(slow,len(numbers)):
    numbers[i] = 0



print(numbers)
```

```python
# Remove all occurrences of value 3 from [3, 2, 2, 3, 4] to get [2, 2, 4]

numbers = [3, 2, 2, 3, 4]

slow = 0

for i in range(len(numbers)):

    if numbers[i] != 3:
        numbers[slow] = numbers[i]
        slow += 1


print(numbers[:slow])
```

```python
# Remove duplicates from SORTED array [1, 1, 2, 2, 3, 4, 4, 5]

numbers = [1, 1, 2, 2, 3, 4, 4, 5]

result = []

seen = set()

for number in numbers:

    if number not in seen:
        seen.add(number)
        result.append(number)

print(result)
```

```python
# Move all EVEN numbers to front, ODD to back: [1, 2, 3, 4, 5, 6]

numbers = [1, 2, 3, 4, 5, 6]

slow = 0

for i in range(len(numbers)):

    if numbers[i] % 2 == 0:
        numbers[slow],numbers[i] = numbers[i],numbers[slow]
        slow += 1

print(numbers)
```

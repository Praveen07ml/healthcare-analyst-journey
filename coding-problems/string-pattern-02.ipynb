# Sring Anagram / Permutation Pattern


```python
# Check if "Hello" and "Olleh" are anagrams (ignore case).

# Approch 1

list1 = "Hello"
list2 = "Olleh"

if len(list1) != len(list2):
    print( "Not Anagram")
    
else:
    sorted1 = ""
    sorted2 = ""
    for i in range(len(list1)):
        sorted1 += list1[i].lower()
        sorted2 += list2[i].lower()

    sorted1 = sorted(sorted1)
    sorted2 = sorted(sorted2)

    if sorted1 == sorted2:
        print("Anagram")
    else:
        print("Not Anagram")

        



# Approach 2 freq type

list1 = "Hello"
list2 = "Olleh"

if len(list1) != len(list2):
    print( "Not Anagram")

else:

    sorted1 = ""
    sorted2 = ""
    for i in range(len(list1)):
        sorted1 += list1[i].lower()
        sorted2 += list2[i].lower()


    freq1 = {}

    for char in sorted1:
        freq1[char] = freq1.get(char,0) + 1

    freq2 = {}

    for char in sorted2:
        freq2[char] = freq2.get(char,0) + 1


    if freq1 == freq2:
        print("Anagram")
    else:
        print("Not Anagram")
    
    
```

```python
# Check if "Hello" and "Olleh" are anagrams (ignore case).

string1 = "Hello"
string2 = "Olleh"


if len(string1) != len(string2):
    print("Anagram")

else:
    string1 = string1.lower()
    string2 = string2.lower()

    sorted1 = sorted(string1)
    sorted2 = sorted(string2)

    if sorted1 == sorted2:
        print("Anagram")
    else:
        print("Not Anagram")
```

```python
# Group anagrams from list: ["eat", "tea", "tan", "ate", "nat", "bat"]

strings = ["eat", "tea", "tan", "ate", "nat", "bat"]

result = {}


for value in strings:
    key = "".join(sorted(value))

    if key not in result:
        result[key] = []
    

    result[key].append(value)
    
print(list(result.values()))
```

```python
# Reverse [1, 2, 3, 4, 5] to get [5, 4, 3, 2, 1]

numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) -1

while left < right:
    numbers[left],numbers[right] = numbers[right],numbers[left]

    left += 1
    right -= 1

print(numbers)
```

```python
# Reverse the string "hello" to get "olleh"

word = "hello" 

result = ""

for i in range(len(word)-1,-1,-1):
    result += word[i]

print(result)
```

```python
# Take last 2 elements + first 3 elements. Use slicing or reversal.
# Expected answer: [4, 5, 1, 2, 3]


numbers = list(range(1,6))

result = []

for i in range(len(numbers)):

    if numbers[i] == 4:
        result.append(numbers[i])
        result.append(numbers[i+1])

for i in range(len(numbers)):
    if numbers[i] < 4:
        result.append(numbers[i])

print(result)
```

```python
# Reverse the words in "hello world python" to get "python world hello"
words = "hello world python"
words = words.split()

result = []

for i in range(len(words)-1,-1,-1):
    result.append(words[i])

print(" ".join(result))
```

```python
# Find the second largest number in [34, 12, 78, 23, 90, 45, 67]

numbers = [34, 12, 78, 23, 90, 45, 67]

if len(numbers) < 2:
    print(None)

else:

    largest = numbers[0]
    second_largest = numbers[1]

    for i in range(len(numbers)):
        number = numbers[i]

        if number > largest:
            second_largest = largest
            largest = number

        elif number > second_largest and number != largest:
            second_largest = number

print(second_largest)
```

```python
# Find the MEDIAN (middle value) of [34, 12, 78, 23, 90]

numbers = [34, 12, 78, 23, 90]
numbers.sort()

answer = []



if len(numbers) % 2 == 0:

    answer.append(len(numbers)//2)

elif len(numbers) % 2 != 0:
    mid = len(numbers)//2

    answer.append(numbers[mid-1]+ numbers[mid]/2)

print(answer)

```

```python
# Remove all zeros from [0, 1, 0, 3, 0, 12] to get [1, 3, 12]

numbers = [0, 1, 0, 3, 0, 12]

for number in numbers[:]:
    if number == 0:
        numbers.remove(number)
print(numbers)
```

```python
# Remove all NEGATIVE numbers from [-5, 3, -2, 8, -1, 10]

numbers = [-5, 3, -2, 8, -1, 10]

result = []

for number in numbers:
    if number > 0 :
        result.append(number)

print(result)
```

```python
# Remove duplicates from [1, 2, 2, 3, 4, 4, 5] to get [1, 2, 3, 4, 5]

numbers = [1, 2, 2, 3, 4, 4, 5]

result = set()

for number in numbers:

    if number not in result:
        result.add(number)

print(result)
```

```python
# Merge [1, 4, 7] and [2, 3, 5, 8, 9] into one sorted list

x = [1, 4, 7]
y = [2, 3, 5, 8, 9]

i = 0
j = 0

result = []

while i < len(x) and j < len(y):

    if x[i] < y[j]:
        result.append(x[i])
        i += 1
    elif x[i] > y[j]:
        result.append(y[j])
        j += 1

while i < len(x):
    result.append(x[i])
    i += 1

while j < len(y):
    result.append(y[j])
    j += 1

print(result)

    
```

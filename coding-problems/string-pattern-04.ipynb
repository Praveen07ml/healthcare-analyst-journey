# String Searching & sub string pattern

```python
# Find all starting positions where "ab" appears in "ababcab"

word = "ababcab"

pattern = "ab"

positions = []

for i in range(len(word) - len(pattern) + 1):
    substring = word[i : i + len(pattern)]

    if pattern == substring:
        positions.append(i)

print(positions)
```

```python
# Check if "hello world" contains the substring "wor"

word = "hello world"

pattern = "wor"

result = False

for i in range(len(word) - len(pattern) + 1):

    substring = word[i : i + len(pattern) ]

    if substring == pattern:
        result = True
        break

print(result)
```

```python
# Count how many times "is" appears in "this is a test, is it?"

words = "this is a test, is it?"

pattern = "is"

count = 0

for i in range(len(words) - len(pattern) + 1):

    substring = words[i : i + len(pattern)]

    if substring == pattern:
        count += 1

print(count)
```

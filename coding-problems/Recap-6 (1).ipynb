```python
# Reverse [1, 2, 3, 4, 5] to get [5, 4, 3, 2, 1]

numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) -1 

while left < right:
    
    numbers[left],numbers[right] = numbers[right],numbers[left]
    left += 1
    right -= 1

print(numbers)
```

```python
# Merge two SORTED lists [1, 3, 5] and [2, 4, 6] into [1, 2, 3, 4, 5, 6]

list1 = [1, 3, 5]
list2 = [2, 4, 6]

i = 0
j = 0

result = []

while i < len(list1) and j < len(list2):

    if list1[i] < list2[j]:
        result.append(list1[i])
        i += 1

    else:
        result.append(list2[j])
        j += 1


while i < len(list1):
    result.append(list1[i])
    i += 1

while j < len(list2):
    result.append(list2[j])
    j += 1

print(result)
    
        
```

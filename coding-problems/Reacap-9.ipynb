```python
# Check if "A man, a plan, a canal: Panama" is a palindrome

sentence = "A man, a plan, a canal: Panama"

words = []
for word in sentence:
    if word.isalpha():
        words.append(word.lower())

words = "".join(words)

left = 0
right = len(words) - 1

result = True

while left < right:
    if words[left] != words[right]:
        result = False
        break

    left += 1
    right -= 1

if result:
    print("Palindrome")

else:
    print("Not Palindrome")

print(words)
```

```python
# Reverse only the VOWELS in "hello" to get "holle"

word = "hello"
word = list(word)

vowels = "aeiouAEIOU"

left = 0
right = len(word) -1

while left < right:

    if word[left] not in vowels:
        left += 1

    elif word[right] not in vowels:
        right -= 1

    else:
        word[left],word[right] = word[right],word[left]
        left += 1
        right -= 1

print("".join(word))
```

```python
# Remove adjacent duplicate characters: "abbaca" → "ca"

# here we are using stack problem 
# think its like a pile of paper plats : you can only pick
# 1st plat or remove it based on condition


word = "abbaca"

stack = []

for char in word:
    if stack and stack[-1] == char: # because of stack[-1] we need also check whether the stack itself empty or not
        stack.pop()

    else:
        stack.append(char)

print("".join(stack))
```

```python
# Compress string "aabcccccaaa" to "a2b1c5a3"

word = "aabcccccaaa"

result = ""

count = 1

for i in range(1,len(word)):

    if word[i] == word[i-1]:
        count += 1

    else:
        result += word[i-1] + str(count)
        count = 1

result += word[-1] + str(count)

print(result)
    
```

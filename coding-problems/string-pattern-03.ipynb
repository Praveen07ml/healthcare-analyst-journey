# String frequency & counting pattern

```python
# Find the FIRST non-repeating character in "aabbccdeef"

word = "aabbccdeef"

count = {}

for char in word:
    count[char] = count.get(char,0) + 1


for key in count:
    if count[key] == 1:
        print(key)
        break
    
```

```python
# Find the MOST COMMON character in "aabbbbcccc"

word = "aabbbbcccc"

count = {}

for char in word:
    count[char] = count.get(char,0) + 1


max_value = max(count.values())

result = []

for key in count:
    if count[key] == max_value:
        result.append(key)

print(result)

```

```python
# Check if all characters in "aabbcc" appear EVEN number of times.

word = "aabbcc"

count = {}

for char in word:
    count[char] = count.get(char,0) + 1

print(count)

result = True

for value in count.values():
    
    if value % 2 != 0:
        result = False
        break

print(result)
```

```python
# Count how many characters appear EXACTLY ONCE in "programming"

word = "programming"

freq = {}

for char in word:
    freq[char] = freq.get(char,0) + 1

print(freq)

count = 0

for value in freq.values():
    if value == 1:
        count += 1

print(count)
```

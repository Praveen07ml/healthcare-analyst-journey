```python
# Check if list [1, 2, 3, 2, 1] is a palindrome (reads same forwards and backwards)

numbers = [1, 2, 3, 2, 1]

left = 0
right = len(numbers) -1

result = True

while left < right:

    if numbers[left] != numbers[right]:
        result = False
        break

    left += 1
    right -= 1

print(result)
```

```python
# In SORTED array [1, 2, 3, 4, 6], find TWO numbers that sum to 6.

numbers = [1, 2, 3, 4, 6]

result = []

left = 0
right = len(numbers) - 1

target = 6

while left < right:

    element = numbers[left] + numbers[right]

    if element == target:
        result.append([numbers[left],numbers[right]])

        left += 1
        right -= 1

    elif element < target:
        left += 1

    else:
        right -= 1

print(result)
```

```python
# Reverse list [1, 2, 3, 4, 5] IN-PLACE using two-pointer swap.

numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) -1

while left < right:

    numbers[left],numbers[right] = numbers[right],numbers[left]

    left += 1
    right -= 1

print(numbers)
```

```python
# Move all zeros to the end: [0, 1, 0, 3, 12] → [1, 3, 12, 0, 0]


numbers = [0, 1, 0, 3, 12]

slow = 0  # for non- zeroes

for fast in range(len(numbers)):

    if numbers[fast] != 0:
        numbers[slow] = numbers[fast]
        slow += 1

for i in range(slow,len(numbers)):
    numbers[i] = 0

print(numbers)
```

```python
# Remove all occurrences of value 3 from [3, 2, 2, 3, 4] to get [2, 2, 4]


numbers = [3, 2, 2, 3, 4]

slow = 0

for fast in range(len(numbers)):

    if numbers[fast] != 3:
        numbers[slow] = numbers[fast]
        slow += 1

print(numbers[:slow])
```

```python
# Remove duplicates from SORTED array [1, 1, 2, 2, 3, 4, 4, 5]

numbers = [1, 1, 2, 2, 3, 4, 4, 5]

result = [numbers[0]]

for i in range(1,len(numbers)):
    if numbers[i] != numbers[i-1]:
        result.append(numbers[i])

print(result)



        
```

```python
# Move all EVEN numbers to front, ODD to back: [1, 2, 3, 4, 5, 6]

numbers = [1, 2, 3, 4, 5, 6]

slow = 0 # for even numbers

for fast in range(len(numbers)):

    if numbers[fast] % 2 == 0:
        numbers[slow],numbers[fast] = numbers[fast],numbers[slow]
        slow += 1

print(numbers)
```

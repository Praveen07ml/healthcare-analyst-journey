#### Understaning Missing Values

```python
print(df.shape)

print(f"\n{df.dtypes}")

print(f"\n{df.isnull().sum()}")
```

```python
total_rows = len(df)

total_desc = df['Description'].isnull().sum()

total_cust = df['CustomerID'].isnull().sum()

print(f"Total Description missing count {total_desc} with {total_desc/total_rows*100:.2f} %")

print(f"Total Customers missing count {total_cust} with {total_cust/total_rows*100:.2f} %")
```

##### 1. Looking into Customer missing values
- i want to separate customers into two data frames one with without missing values for `customer analysis`
- one with all customers with `revenue analaysis`

```python
df_with_customers = df[df['CustomerID'].notna()].copy()
print(f"total_customers with ids {len(df_with_customers)}")

df_with_all_customers = df.copy()
print(f"total_customers {len(df_with_all_customers)}")
```

##### 2. Looking into Negative Qauntities

```python
negative_quantities = df[df['Quantity'] < 0]

print(f"Negative quantities rows {len(negative_quantities)} with {len(negative_quantities)/len(df)*100:.2f} %\n ")


negative_quantities[['InvoiceNo','StockCode','Description','Quantity','InvoiceDate','UnitPrice','CustomerID']].head()


negative_quantities['InvoiceNo'].str[0].head(10)
```

```python
df_sales = df[df['Quantity'] > 0].copy()
df_returns = df[df['Quantity'] < 0].copy()

print(f"Sales transactions {len(df_sales)}")
print(f"Sales return transactions {len(df_returns)}")


df_sales['Revenue'] = df_sales['Quantity'] * df_sales['UnitPrice']
print(f"\nTotal revenue {df_sales['Revenue'].sum():,.2f}$\n")

df_sales.head()
```

#### Understanding Missing Values

```python
total_rows = len(df)

missing_customer = df['CustomerID'].isnull().sum()

missing_description = df['Description'].isnull().sum()

print(f"Total Rows: {total_rows}")
print(f"Missing CustomerID {missing_customer} rows with ({missing_customer/total_rows*100:.1f}%)")
print(f"Missing Description {missing_description} rows with ({missing_description/total_rows*100:.1f}%)")

```

- Looking at Missing Customer IDs
- i Would separate `two different DataFrames`
- one with Customer IDs for `Customer Analysis`
- one with all rows for `Revenue Analysis`
- I do not drop them because its still represent the Whole transactions with real revenue

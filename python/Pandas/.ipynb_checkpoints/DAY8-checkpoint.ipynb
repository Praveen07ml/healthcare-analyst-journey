## Business Questions based on columns

#### 1. ID/Key Columns

```python
print(f"Unique Invoice No - {df['InvoiceNo'].nunique()}")

print(f"Unique Stock code - {df['StockCode'].nunique()}")

print(F"Unique Customer IDs - {df['CustomerID'].nunique()}")


# Customer purchasing the no of orders 

print(f"Per customer purchase count  {df['InvoiceNo'].nunique() / df['CustomerID'].nunique()}")
```

#### 2. Category / Grouping Columns 

```python
print("Country Segmentation\n")

print(df['Country'].value_counts().head(10))
```

#### 3. Subject Columns

```python
print("Product unique counts\n")

df['Description'].value_counts().head(10)
```

#### 4. Numeric Measure columns

```python
print(f"Avg of Unit Price - {df['UnitPrice'].mean():.2f}")
print(f" Total Quantity - {df['Quantity'].sum()}")
```

#### 5. Date columns

```python
print(f" This dataset Date range between {df['InvoiceDate'].min()} from {df['InvoiceDate'].max()}")
```

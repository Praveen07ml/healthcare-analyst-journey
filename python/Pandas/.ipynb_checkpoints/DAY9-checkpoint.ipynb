#### Recoring Measures

```python
print(f"Initial shape {df.shape}\n")

print(f" Data columns {df.columns}\n")

print(f" total Missing values {df.isnull().sum()}")
```

### Looking into missing values 

```python
print(f" Total rows {len(df)}\n")

print(f"Missing customers ids - {df['CustomerID'].isnull().sum()} with {df['CustomerID'].isnull().sum()/len(df):.1f} % \n")

print(f"Missing Descriptions - {df['Description'].isnull().sum()} with {df['Description'].isnull().sum() / len(df):.1f} %")

```

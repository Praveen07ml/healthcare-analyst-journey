### Looking into missing Values

```python
total_rows = len(df)
missing_customers = df['CustomerID'].isnull().sum()
missing_products = df['Description'].isnull().sum()

print(f"Total rows {total_rows}")
print(f"Missing Customers {missing_customers} rows with {missing_customers/total_rows*100:.2f} %")
print(f"Missing Products {missing_products} rows with {missing_products/total_rows*100:.2f} %")
```

#### Looking inot customerIDs
- i will create two separate Dataframes one with only customers for customer analysis and another dataframe with all rows for Revenue analysis
- becuase the given transactions still represent original transactions with actual revenue

```python
df[df['CustomerID'].isnull()].head()
```

```python
df_with_customers = df[df['CustomerID'].notna()].copy()
with_all_transactions = df.copy()

print(f"Customers with customerIDs {len(df_with_customers)} rows")
print(f"Customers with all transactions {len(with_all_transactions)} rows")
```

#### Looking Negative Quanitites

```python
negative_rows = df[df['Quantity'] < 0]

print(f"Total Negative rows {len(negative_rows)}")
print(f"Percentage of Negative rows {len(negative_rows)/len(df)*100:.1f} %\n")


print("Sample Negative Rows\n")

print(negative_rows[['InvoiceNo','StockCode','Description','Quantity']].head(10))

print("\nDo Negative invoices start with special characters\n")


negative_rows['InvoiceNo'].str[0].value_counts()
```

```python
df_sales = df[df['Quantity']>0].copy()
df_returns = df[df['Quantity']<0].copy()

print(f"Total Sales Transactions {len(df_sales)}")
print(f"Total Sales Returns transactions {len(df_returns)}")

df_sales['Revenue'] = df_sales['Quantity'] * df_sales['UnitPrice']

print("\n Total Sales Revenue")
print(f"\n$ {df_sales['Revenue'].sum():,.2f}")

```
